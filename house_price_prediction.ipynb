# House Price Prediction with CatBoost

This notebook implements a machine learning model to predict house prices using the CatBoostRegressor algorithm.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score
from catboost import CatBoostRegressor
```

## 2. Load Dataset

```python
data = pd.read_csv("house_prices.csv")

print(data.head())
print(data.shape)
print(data.info())
```

## 3. Data Preprocessing

```python
# Convert categorical column to numerical columns
data = pd.get_dummies(data, columns=["City"])

# Separate features and target
X = data.drop("Price", axis=1)
y = data["Price"]
```

## 4. Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.1,
    random_state=42
)
```

## 5. Create Model

```python
model = CatBoostRegressor(
    save_snapshot=True,
    snapshot_file="Test_house_price",
    snapshot_interval=2,
    verbose=False
)
```

## 6. Train Model

```python
model.fit(X_train, y_train)
```

## 7. Prediction

```python
pred = model.predict(X_test)
```

## 8. Model Evaluation

```python
mae = mean_absolute_error(y_test, pred)
r2 = r2_score(y_test, pred)

print("MAE:", mae)
print("R² Score:", r2)
```
